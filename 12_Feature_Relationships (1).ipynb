## **1. Correlation Matrix**



**Understand the Concept**

* Correlation matrix shows the correlation between multiple variables.
* It helps us see which variables are related to each other.
* The value is between -1 and +1.
* +1 - strong positive correlation.
* -1 - strong negative correlation.
* 0 - little or no linear relationship.



**Demonstrate the Concept**

**Example:**

* Study Hours, Exam Marks, and Attendance are three variables.
* We can calculate the correlation between every pair of variables.
* This gives us a correlation matrix.



```python
# Labraries
import pandas as pd
import numpy as np
```

```python
# Implement the Concept:
# Student data
data = pd.DataFrame({"Study_Hours": [2, 4, 6, 8, 10],
                     "Attendance": [60, 70, 75, 85, 95],
                     "Exam_Marks": [50, 60, 70, 80, 90]})

# correlation matrix
correlation_matrix = data.corr()

print(correlation_matrix)
```

## **2. Multicollinearity**



**Understand the Concept**

* Multicollinearity happens when two or more features are strongly related to each other.
* Highly related features can provide almost the same information.



**Demonstrate the Concept**

**Example:**

* A house dataset has House Area and Number of Rooms.
* Usually, a bigger house has more rooms.
* So, these two features may be strongly related.





```python
# Implement the Concept:
# House data
data = pd.DataFrame({"House_Area": [800, 1000, 1200, 1500, 1800],
                     "Number_of_Rooms": [2, 2, 3, 4, 5]})

# correlation 
correlation = data.corr()

print(correlation)
```

## **3. Variance Inflation Factor (VIF)**



**Understand the Concept**

* VIF is used to check multicollinearity between features.
* It tells us how much one feature is related to the other features.
* A high VIF means the feature has a strong relationship with other features.
* A VIF close to 1 usually means there is little multicollinearity.

**Formula :** VIF = 1 / (1 - R²)

* R² shows how well the other features can explain the selected feature.



**Demonstrate the Concept**

**Real-Life Example:**

* A house dataset contains house area, number of rooms, and number of bedrooms.
* Some of these features may give similar information.
* VIF helps us find features that are too closely related.


```python
# Implement the Concept:
from statsmodels.stats.outliers_influence import variance_inflation_factor

# House data
data = pd.DataFrame({"House_Area": [800, 1000, 1200, 1500, 1800],
                     "Number_of_Rooms": [2, 2, 3, 4, 5],
                     "Bedrooms": [1, 1, 2, 3, 4]})

# VIF
vif = pd.DataFrame() # Empty dataset
# Add 2 col
vif["Feature"] = data.columns
vif["VIF"] = [variance_inflation_factor(data.values, i)for i in range(data.shape[1])]

print(vif)
```

## **4. Feature Importance**



**Understand the Concept**

* Feature importance tells us which features are more useful for making predictions.
* Important features have more influence on the model's prediction.



**Demonstrate the Concept**

**Example:**

* Suppose we want to predict house prices.
* Features are House Area, Number of Rooms, and Age of House.
* House Area may have more importance because it has a strong effect on the house price.



```python
# Implement the Concept:
from sklearn.ensemble import RandomForestRegressor

# House data
data = pd.DataFrame({"House_Area": [800, 1000, 1200, 1500, 1800, 2000],
                     "Number_of_Rooms": [2, 2, 3, 4, 5, 5],
                     "Age_of_House": [15, 12, 10, 8, 5, 3],
                     "Price": [30, 35, 45, 55, 70, 80]})

# Features and target
X = data[["House_Area", "Number_of_Rooms", "Age_of_House"]]
y = data["Price"]

# Train model
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Feature importance
importance = pd.Series(model.feature_importances_,index=X.columns)

print(importance)
```

## **5. Feature Selection Basics**



**Understand the Concept**

* Feature selection means choosing the most useful features from a dataset.
* We remove features that are not useful or give repeated information.
* It helps make the ML model simpler and faster.
* Good feature selection can also improve model performance.



**Demonstrate the Concept**

**Example:**

* Suppose we want to predict house prices.
* We have features like House Area, Number of Rooms, House Color, and House Price.
* House Area and Number of Rooms may be useful.
* House Color may not help much in predicting the price.
* So, we can select the useful features and remove the unnecessary one.



```python
# Implement the Concept:
from sklearn.feature_selection import SelectKBest, f_regression

# House price data
data = pd.DataFrame({"House_Area": [800, 1000, 1200, 1500, 1800, 2000],
                     "Number_of_Rooms": [2, 2, 3, 4, 5, 5],
                     "Age_of_House": [15, 12, 10, 8, 5, 3],
                     "House_Color": [1, 2, 1, 3, 2, 1],
                     "Price": [30, 35, 45, 55, 70, 80]})

# Features and target
X = data[["House_Area","Number_of_Rooms","Age_of_House", "House_Color"]]

y = data["Price"]

# Select the 2 best features
selector = SelectKBest(score_func=f_regression, k=2)
X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]

print("Selected Features:", list(selected_features))
```

This method checks which features have the strongest relationship with the house price.
