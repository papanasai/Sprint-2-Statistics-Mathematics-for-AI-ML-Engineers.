# What is Statistics?

```python
# Correct list of sales as integers
soft_drink_sales = [20000, 40000, 60000]
# Calculate average
average = sum(soft_drink_sales) / len(soft_drink_sales)
# Print result
print("Average soft drink sales:", average)
```

## Interpretation
## The average soft_drink sales is 40000
