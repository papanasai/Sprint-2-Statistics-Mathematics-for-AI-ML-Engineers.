## **1. Sampling**

**Understand the Concept**

* Sampling means selecting a small group from a larger population.
* To collect information and draw conclusions about the entire population.

**Demonstrate the Concept**

**Example:**

* A college has 1,000 students.
* We select 100 students for a survey.
* These 100 students are the sample.
* The 1,000 students are the population.


```python
# Labraries:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Implement the Concept:
# Population - 1000 students
population = np.arange(1, 1001)

# Sample - 100 students - choose random values in population
sample = np.random.choice(population, size=100, replace=False)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("Sample Students:", sample[:10])
```

## **2. Random Sampling**

**Understand the Concept**

* Random sampling means selecting people or data randomly from a population.
* Every member has an equal chance of being selected.



**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to know how employees feel about a new workplace policy.
* It randomly selects 50 employees from the full employee list.
* This gives different employees an equal chance to take part in the survey.



```python
# Implement the Concept:
import random
# Population
employees = list(range(1,501))
# Random sample
random_employees = random.sample(employees,50)
print(random_employees[30:]) # Slicing
```

## **3. Stratified Sampling**

**Understand the Concept**

* Stratified sampling means dividing a population into different groups.
* These groups are called strata.
* We then select samples from each group.

**Demonstrate the Concept**

**Example:**

* A college has 600 students.
* There are 300 in 1st year students, 200 in 2nd year students, and 100 in 3th year students.
* We can select 30 from 1st year, 20 from 2nd year, and 10 from 3th year, total 60 samples.


```python
# Implement the Concept:
# Population
college_students = list(range(1,601))
# Groups
first_year = college_students[:301]
second_year = college_students[301:501]
third_year = college_students[501:]
# Sample
sample = (random.sample(first_year,30) + 
          random.sample(first_year,20) +
          random.sample(first_year,10))
print(sample[:25])
```

## **4. Cluster Sampling**

**Understand the Concept**

* Cluster sampling means dividing a population into groups called clusters.
* Instead of selecting people from every group, we select some complete groups.
* Then we collect data from all or many members of the selected groups.



**Demonstrate the Concept**

**Real-Life Example:**

* A company has offices in 3 cities.
* Instead of surveying employees from every city, it selects 1 cities.
* All employees from those selected cities are surveyed.




```python
# Implement the Concept:
# Population
employees = pd.DataFrame({"Name": ["Kavin", "Pravin", "Navin", "Ashwin", "Pavi",
                                   "Sai", "Prabha", "Jevi", "Kanna"],
                          "City": ["Chennai", "Chennai", "Chennai","Coimbatore",
                                   "Coimbatore", "Coimbatore","Bengaluru",
                                   "Bengaluru", "Bengaluru"]})

# Groups - randomly select 2 cities
selected_cities = np.random.choice(employees["City"].unique(),size=2,replace=False)

# Sample - selected city employees
sample = employees[employees["City"].isin(selected_cities)]

print("Selected Groups:", selected_cities)
print("Sample: \n", sample)
```

## **5. Systematic Sampling**



**Understand the Concept**

* Systematic sampling means selecting data at regular intervals from a population.
* We choose a starting point and then select every fixed number of members.
* The fixed number is called the sampling interval.



**Demonstrate the Concept**

**Numerical Example:**

* A company has 100 employees.
* We want to select 10 employees.
* Sampling interval = 100 / 10 = 10.
* Starting from the 5th employee, we select every 10th employee.
* So, employees 5, 15, 25, 35, and so on are selected.





```python
# Implement the Concept:
# Population - 100 employees
employees = np.arange(1, 101)

# Systematic sample 
systematic_sample = employees[4::10] # start:stop:step - slicing
print("Selected Employees:", systematic_sample)
```

## **6. Sampling Bias**

**Understand the Concept**

* Sampling bias happens when the selected sample does not properly represent the whole population.
* It can give misleading or unfair results.
* It usually happens when some groups are more likely to be selected than others.

**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to know whether employees like working from home.
* If the survey is given only to employees who already work from home, the result may be biased.


```python
# Implement the Concept:
# Population: 80 office employees and 20 remote employees
employees = np.arange(1,101)

# Groups - 80 WFO & 20 WFH employees
WFO_employees = employees[:81]
WFH_employees = employees[81:]

# Biased sample -  selecting only WFO employees
biased_sample = np.random.choice(WFO_employees, size=20, replace=False)

print("Population Size:", len(employees))
print("Biased Sample Size:", len(biased_sample))
print("WFO Employees in Sample:", len(biased_sample))
print("WFH Employees in Sample:", 0)
```

**Explaination**
* The population has both office and remote employees.
* The sample contains only office employees.
* Remote employees are not represented in the sample.
* So, the sample can give a biased result.

## **7. Sampling Error**

**Understand the Concept**

* Sampling error is the difference between the result from a sample and the actual result from the whole population.
* It happens because we study only a part of the population

**Demonstrate the Concept**

**Example:**

* A school has 1,000 students.
* The actual average mark of all students is 75.
* We select 100 students and their average mark is 72.
* The difference of 3 marks is called sampling error.

```python
# Implement the Concept:

# Population - Marks
population_marks = np.array([70, 72, 74, 75, 76, 78, 80, 82, 73, 75])

# Sample - Selected values
sample_marks = np.array([70, 72, 74, 75, 76])

# Mean - population & sample 
population_mean = np.mean(population_marks)
sample_mean = np.mean(sample_marks)

# Sampling error
sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
```

## **8. Sample Size**

**Understand the Concept**

* Sample size means the number of observations or people selected from a population.
* The sample should be large enough to give a useful result.
* A very small sample may not represent the population well.
* A larger sample usually gives more reliable results, but it may take more time and resources.

**Demonstrate the Concept**

**Example:**

* A company has 10,000 customers.
* It wants to conduct a customer satisfaction survey.
* It selects 500 customers for the survey.
* So, the sample size is 500.


```python
# Implement the Concept:
# Population - 10000 customers
customers = np.arange(1, 10001)

# Sample -  500 customers - 5%
sample = np.random.choice(customers, size=500, replace=False)

print("Population Size:", len(customers))
print("Sample Size:", len(sample))
```
