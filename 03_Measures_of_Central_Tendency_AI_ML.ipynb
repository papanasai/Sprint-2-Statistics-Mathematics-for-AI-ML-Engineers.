## 1. Mean

### Definition
The mean is the arithmetic average of all observations.

### Formula
**Mean = Sum of all values / Number of values**

### Numerical Example
Data: `10, 20, 30, 40, 50`

Mean = `(10 + 20 + 30 + 40 + 50) / 5 = 30`


```python
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)

print("Mean:", mean)
```

### Business Example
Monthly sales are ₹10L, ₹12L, ₹15L, ₹13L, and ₹20L.

The average monthly sales are ₹14L.

### AI/ML Example
For customer ages `[20, 25, 30, 35, 40]`, the mean helps describe the average customer age during EDA.

### Interpretation
Mean is useful when data does not contain strong outliers. It is sensitive to extreme values.


## 2. Median

### Definition
The median is the middle value after sorting the data.

### Formula
- Odd number of observations: middle value
- Even number: average of the two middle values

### Numerical Example
Data: `10, 20, 30, 40, 50`

Median = `30`


```python
import statistics

data = [10, 20, 30, 40, 50]

median = statistics.median(data)

print("Median:", median)
```

### Business Example
Employee salaries are ₹20K, ₹25K, ₹30K, ₹35K, and ₹2L.

The median is ₹30K and is less affected by the very high salary.

### AI/ML Example
House prices such as `[40, 45, 50, 55, 300]` have an extreme outlier. The median better represents a typical house price.

### Interpretation
Median is useful for skewed data and datasets containing outliers.


## 3. Mode

### Definition
The mode is the value that occurs most frequently.

### Numerical Example
Data: `10, 20, 20, 30, 40`

Mode = `20`


```python
import statistics

data = [10, 20, 20, 30, 40]

mode = statistics.mode(data)

print("Mode:", mode)
```

### Business Example
A clothing store sells sizes `M, L, M, S, M, XL, L`.

The most frequently sold size is M.

### AI/ML Example
For payment methods `UPI, Card, UPI, Cash, UPI, Card`, the mode is UPI.

### Interpretation
Mode is especially useful for categorical data and identifying the most common category.


## 4. Weighted Mean

### Definition
A weighted mean gives different importance to different observations.

### Formula
**Weighted Mean = Σ(wx) / Σw**

### Numerical Example
Marks = `[80, 70, 90]`
Weights = `[3, 2, 5]`

Weighted Mean = `(80×3 + 70×2 + 90×5) / (3+2+5) = 83`


```python
marks = [80, 70, 90]
weights = [3, 2, 5]

weighted_mean = sum(
    mark * weight for mark, weight in zip(marks, weights)
) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

### Business Example
Employee evaluation:
- Technical Skills = 80, weight 50%
- Communication = 70, weight 20%
- Performance = 90, weight 30%

Weighted score = 82.

### AI/ML Example
Weighted calculations can be used when samples have different importance, such as weighted errors or sample-weighted model training.

### Interpretation
Use weighted mean when some observations contribute more than others.


## 5. Geometric Mean

### Definition
The geometric mean is the nth root of the product of n positive values.

### Formula
**GM = (x₁ × x₂ × ... × xₙ)^(1/n)**

### Numerical Example
Data: `2, 4, 8`

GM = cube root of `(2 × 4 × 8)` = `4`


```python
import math

data = [2, 4, 8]

geometric_mean = math.prod(data) ** (1 / len(data))

print("Geometric Mean:", geometric_mean)
```

### Business Example
Revenue growth factors over three years are `1.10, 1.20, 1.30`.

Geometric mean summarizes the typical multiplicative growth factor.

### AI/ML Example
Performance factors such as `[1.05, 1.10, 1.20]` can be summarized with a geometric mean when changes are multiplicative.

### Interpretation
Geometric mean is useful for growth rates, investment returns, ratios, and multiplicative changes. Standard geometric mean requires positive values.


```python
growth_factors = [1.10, 1.20, 1.30]

gm = math.prod(growth_factors) ** (1 / len(growth_factors))

print("Average Growth Factor:", gm)
print("Approximate Annual Growth Rate:", (gm - 1) * 100, "%")
```

## 6. Harmonic Mean

### Definition
The harmonic mean is the reciprocal of the arithmetic mean of reciprocals.

### Formula
**HM = n / Σ(1/x)**

### Numerical Example
For equal-distance speeds of 60 km/h and 40 km/h:

HM = `2 / (1/60 + 1/40) = 48 km/h`


```python
import statistics

speeds = [60, 40]

harmonic_mean = statistics.harmonic_mean(speeds)

print("Harmonic Mean:", harmonic_mean)
```

### Business Example
For comparable delivery routes with speeds of 60 km/h and 40 km/h, the harmonic mean is 48 km/h.

### AI/ML Example — F1 Score
F1-score is the harmonic mean of precision and recall.

If:
- Precision = 0.80
- Recall = 0.60

F1 = `2 × 0.80 × 0.60 / (0.80 + 0.60) = 0.6857`


```python
precision = 0.80
recall = 0.60

f1_score = (2 * precision * recall) / (precision + recall)

print("F1 Score:", f1_score)
```

### Interpretation
Harmonic mean is useful for rates and ratios. In AI/ML, F1-score balances precision and recall and becomes lower when either one is low.
