# Hypothesis Testing**

Hypothesis testing is a method used to check whether there is enough evidence to support a claim about data.

## **1. Null Hypothesis**



**Understand the Concept**

* Null hypothesis says there is no change, no difference, or no relationship.
* It is usually written as **H₀**.
* We start a hypothesis test by assuming the null hypothesis is true.



**Demonstrate the Concept**


**Real-Life Example:**

* A company introduces a new training program.
* We want to check whether it changes employee performance.
* The null hypothesis is company assumes the average employee score is 75.

**AI/ML Example:**

* In Machine Learning experiments, the null hypothesis can be used to check whether a new feature or method actually improves model performance.



```python
# Libraries
import numpy as np
import pandas as pd
```

```python
# Implement the Concept:
# Employee performance scores 
scores = np.array([72, 75, 78, 74, 76, 73, 77, 75, 79, 76])

# Null Hypothesis - avg score = 75
null_hypothesis = scores.mean()

print("Null Hypothesis:", null_hypothesis)
```

## **2. Alternative Hypothesis**



**Understand the Concept**

* Alternative hypothesis says there is a change, difference, or relationship.
* It is usually written as **H₁** or **Hₐ**.
* It is considered when there is enough evidence to reject the null hypothesis.



**Demonstrate the Concept**

**Example:**

* A company says the average delivery time is 30 minutes.
* We want to check whether the actual average delivery time is different from 30 minutes.
* **H₀:** Average delivery time = 30 minutes.
* **H₁:** Average delivery time ≠ 30 minutes.


```python
# Implement the Concept:
# Delivery times data
delivery_time= np.array([25,30,35,40,42,20])

# H0: Average delivery time is 30 minutes.
# H1: Average delivery time is not 30 minutes.

# Avg delivery time(Mean)
avg_delivery_time = delivery_time.mean()
print("Avg delivery time", avg_delivery_time)
```

* H0 : 30
* H1 : 32
* Here H1 Alternative Hypothesis is correct.

## **3. p-value**

**Understand the Concept**

* p-value tells us how strong the evidence is against the null hypothesis.
* A small p-value means there is strong evidence against the null hypothesis.
* A large p-value means there is not enough evidence to reject the null hypothesis.
* A common value used for checking is 0.05.





**Demonstrate the Concept**

**Real-Life Example:**

* You normally reach college in 30 minutes.
* One day, you try a different route and reach in 25 minutes.
* We can use a statistical test and p-value to check whether the new route really saves time or the difference happened by chance.

**AI/ML Example:**

* p-value can help check whether a feature has a meaningful relationship with the target value.
* It is also used in statistical tests during data analysis.


```python
import numpy as np
from scipy.stats import ttest_1samp

# Travel time
travel_times = np.array([
    26, 25, 24, 27, 25,
    26, 24, 25, 26, 24
])

# H0: Average travel time is 30 minutes
# H1: Average travel time is not 30 minutes

# One-sample t-test
result = ttest_1samp(
    travel_times,
    popmean=30
)

print("T-statistic:", result.statistic)
print("p-value:", result.pvalue)

if result.pvalue < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **4. Confidence Interval**



**Understand the Concept**

* Confidence interval gives a range of values where the actual population value is likely to be.
* It helps us understand how much uncertainty is in a sample result.
* A 95% confidence interval is commonly used.



**Demonstrate the Concept**

**Example:**

* A person checks the time taken to reach college.
* The average travel time from a sample is 25 minutes.
* A 95% confidence interval may be from 24 to 26 minutes.
* This gives a range for the actual average travel time.



```python
# Implement the Concept:
import numpy as np
# Travel times 
times = np.array([24, 25, 26, 25, 24, 27, 25, 26, 24, 25])

# Avg travel time
mean = np.mean(times)


print("Average Travel Time:", mean)
print("95% Confidence Interval is 24 to 26")
```

## **5. Significance Level**



**Understand the Concept**

* Significance level tells us the limit we use to decide whether to reject the null hypothesis.
* It is usually written as α (alpha).
* A common significance level is 0.05.
* If the p-value is less than 0.05, we reject the null hypothesis.



**Demonstrate the Concept**

**Real-Life Example:**

* A student wants to check whether a new study method improves test scores.
* The student chooses a significance level of 0.05 before doing the test.



```python
# Implement the Concept:
# Test scores 
scores = np.array([72, 75, 78, 74, 76, 77, 73, 79, 75, 78])

# H0: Average score = 75
# H1: Average score != 75
result = ttest_1samp(scores, popmean=75)

print("p-value:", result.pvalue)
print("Significance Level:", alpha)

# Significance level
alpha = 0.05
if result.pvalue < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **6. Type I Error**



**Understand the Concept**

* Type I Error happens when we reject the null hypothesis even though it is actually true.
* It is also called a **false positive**.



**Demonstrate the Concept**

**Example:**

* A person is not sick.
* A medical test says the person is sick.
* The test result is wrong.
* This is a Type I Error.




```python
print("Type I Error: Rejecting a true Null Hypothesis")
```

## **7. Type II Error**



**Understand the Concept**

* Type II Error happens when we do not reject the null hypothesis even though it is actually false.
* It is also called a false negative.



**Demonstrate the Concept**

**Example:**

* A person actually has a disease.
* The medical test says the person does not have the disease.
* The test result is wrong.
* This is a Type II Error.



```python
print("Type II Error: Failing to reject a false Null Hypothesis")
```

## **8. One-tailed Test**



**Understand the Concept**

* A one-tailed test checks for a difference in only one direction.
* We check whether a value is greater than or less than a particular value.




**Demonstrate the Concept**

**Real-Life Example:**

* A person buys a new fast charger.
* They want to check whether the new charger charges the phone in less time than the old charger.
* Since we are checking only for a decrease, this is a one-tailed test.

```python
# Implement the Concept:
# New Charger charging times 
charging_times = np.array([110, 115, 112, 108, 114, 111, 109, 113, 110, 112])

# H0: Average charging time = 120 minutes
# H1: Average charging time less than 120 minutes 

# T-test 
result = ttest_1samp(charging_times, popmean=120)

# p_value : p-value / 2 - why means it return two-tailed
p_value = result.pvalue / 2

print("Average Charging Time:", np.mean(charging_times))
print("One-tailed p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **9. Two-tailed Test**



**Understand the Concept**

* A two-tailed test checks for a difference in both directions.
* It checks whether a value is greater than or less than a given value.



**Demonstrate the Concept**

**Real-Life Example:**

* A student changes their study method.
* They want to know whether the new method changes their exam marks.
* The marks could increase or decrease.
* So, a two-tailed test is suitable.



```python
# Implement the Concept:
# Exam marks 
marks = np.array([72, 75, 78, 74, 76, 73, 77, 75, 79, 76])

# H0: Average mark = 75
# H1: Average mark != 75

result = ttest_1samp(marks, popmean=75)

print("Average Mark:", np.mean(marks))
print("Two-tailed p-value:", result.pvalue)

if result.pvalue < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```
