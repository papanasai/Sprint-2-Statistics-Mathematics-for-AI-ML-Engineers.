# 4. Detecting Outliers

# IQR Method

# Concept for Understanding
IQR stands for Interquartile Range.                   IQR = Q3-Q1
Where:
- Q1 = 25th percentile
- Q2 = Median / 50th percentile
- Q3 = 75th percentile
Outlier boundaries
Lower\ Bound = Q1-1.5(IQR)
Upper\ Bound = Q3+1.5(IQR)
Any value outside these boundaries is generally considered an outlier.

# Numerical Example
10, 12, 13, 14, 15, 16, 18, 20, 100

```python
# Python — IQR Method
import numpy as np
data = np.array([
    10, 12, 13, 14, 15,
    16, 18, 20, 100
])
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

# Box Plot
A box plot is one of the most common visual methods for identifying outliers.
It displays:
- Minimum/non-outlier lower value
- Q1
- Median
- Q3
- Maximum/non-outlier upper value
- Potential outliers

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Employee": ["A", "B", "C", "D", "E", "F", "G"],
    "Salary": [25000, 30000, 35000, 40000, 45000, 50000, 150000]
}
df = pd.DataFrame(data)
print(df)
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df["Salary"])
plt.ylabel("Salary")
plt.title("Salary Box Plot")
plt.show()
```

# Scatter Plot
A scatter plot is particularly useful for identifying unusual observations between two variables.

```python
# Example:
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Experience": [1, 2, 3, 5, 7, 10],
    "Salary": [25000, 30000, 35000, 45000, 60000, 90000]
}
df = pd.DataFrame(data)
plt.scatter(df["Experience"], df["Salary"])
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```
