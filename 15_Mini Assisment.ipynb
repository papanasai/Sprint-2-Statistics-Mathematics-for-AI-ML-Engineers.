# 1. Explain the difference between Mean and Median.
Mean → Average of all values, calculated by summing all values and dividing by the count.
Median → The middle value once the data is sorted.
The mean is affected by outliers, while the median is more robust to extreme values

```python
data = [10, 12, 14, 15, 100]
# Mean
mean_value = np.mean(data)
# Median
median_value = np.median(data)
print("Mean:", mean_value)
print("Median:", median_value)
```

# 2. What is Variance?
Variance measures how far data values are spread from the mean. A higher variance indicates greater variability.

```python
data = [10, 12, 23, 23, 16, 23, 21, 16]
mean = sum(data) / len(data)
variance = sum((x - mean) ** 2 for x in data) / len(data)  # population variance
print("Variance:", variance)
```

# 3. What is Standard Deviation?
Standard deviation is the square root of variance. It shows the average distance of data points from the mean, expressed in the same unit as the data.

```python
import statistics as stats
data = [10, 12, 23, 23, 16, 23, 21, 16]
std_dev = stats.pstdev(data)  # population standard deviation
print("Standard Deviation:", std_dev)
```

# 4. Explain Bayes' Theorem with an example.
Bayes' Theorem calculates the probability of an event based on prior knowledge of related conditions.

Formula: P(A|B) = [P(B|A) * P(A)] / P(B)

Example: A medical test estimates the probability that a person has a disease after receiving a positive test result, combining the test's accuracy with how rare the disease is.

```python
P_D = 0.01          # Probability of disease
P_not_D = 1 - P_D   # Probability of no disease

P_pos_given_D = 0.95      # True positive rate
P_pos_given_not_D = 0.05  # False positive rate

# Total probability of testing positive
P_pos = P_pos_given_D * P_D + P_pos_given_not_D * P_not_D

# Bayes' Theorem: Probability of disease given positive test
P_D_given_pos = (P_pos_given_D * P_D) / P_pos

print("Probability of having the disease after a positive test:", P_D_given_pos)
```

# 5. Difference between Population and Sample.
Population:
Entire group under study
Usually very large
Example: All employees in a company
Sample:
Subset of the population
Smaller and easier to analyze
Example: 50 employees selected for analysis

```python
import random
# Population: employee IDs 1 to 500
population = list(range(1, 501))
# Sample: randomly select 50 employees
sample = random.sample(population, 50)
print("Population size:", len(population))
print("Sample size:", len(sample))
print("Sample:", sample)
```

# 6. Explain Correlation and Covariance.
Covariance indicates the direction of the linear relationship between two variables (positive or negative), but its magnitude is not standardized.
Correlation standardizes covariance to a range of -1 to +1, making it easier to interpret the strength and direction of the relationship.

```python
import numpy as np
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
# Covariance matrix
cov_matrix = np.cov(x, y)
cov_xy = cov_matrix[0, 1]
# Correlation matrix
corr_matrix = np.corrcoef(x, y)
corr_xy = corr_matrix[0, 1]

print("Covariance:", cov_xy)
print("Correlation:", corr_xy)
```

# 7. What is Skewness?
Skewness measures the asymmetry of a distribution.
Positive Skew → Long right tail
Negative Skew → Long left tail
Zero Skew → Symmetric distribution

```python
from scipy.stats import skew
import numpy as np
data = np.array([1, 2, 2, 3, 4, 10])
skewness_value = skew(data)
print(skewness_value)
```

# 8. Explain IQR.
IQR (Interquartile Range) is the difference between the third quartile (Q3) and the first quartile (Q1). It measures the spread of the middle 50% of the data and is commonly used to detect outliers.

```python
import numpy as np
data = [10, 12, 14, 15, 18, 21, 22, 25, 30]
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
print("IQR:", IQR)
```

# 9. What is a p-value?
The p-value is the probability of observing the obtained result if the null hypothesis is true.
p-value < 0.05 → Reject the Null Hypothesis
p-value ≥ 0.05 → Fail to Reject the Null Hypothesis

```python
from scipy import stats
import numpy as np
data = np.array([5.1, 4.9, 5.3, 5.0, 5.2])
mu0 = 5  # null hypothesis mean
t_stat, p_value = stats.ttest_1samp(data, mu0)
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

# 12. Detect Outliers using the IQR Method.

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["Salary"] < lower) | (df["Salary"] > upper)]
outliers
```
