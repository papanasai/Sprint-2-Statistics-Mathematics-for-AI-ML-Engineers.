# Correlation & Covariance

# 1. Correlation
Concept Explanation
Correlation measures the strength and direction of the relationship between two variables.

# Real-World Example
Exercise time and calories burned
Generally, as exercise time increases, calories burned also increase.

# Business Example
Advertising expenditure and sales
Higher advertising expenditure may be associated with higher sales.

# AI/ML Example
Before training a house-price model, we can examine:
Area ↔ House Price
Bedrooms ↔ House Price
Age ↔ House Price

# 2. Covariance
Concept Explanation
Covariance measures the direction in which two variables change together.

# Interpretation
   Covariance	                          Meaning
    Positive	             X and Y tend to increase together
    Negative	           One increases while the other tends to decrease
    Near 0	               Little linear relationship

# Real-World Example
Temperature and ice-cream sales generally have positive covariance.

# Business Example
Marketing spending and sales may have positive covariance.

# AI/ML Example
Covariance can be used during EDA to understand how numerical features move together.

```python
import pandas as pd
df = pd.DataFrame({
     "Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 45, 55, 65, 75]
})
print(df.cov(numeric_only=True))
```

# 6. Pearson Correlation

# understanding the Concept Explanation
Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.

# Formula
- r = Pearson correlation coefficient
- Cov(X,Y) = covariance
- sigma_X = standard deviation of X
- sigma_Y = standard deviation of Y

```python
pearson = df["Hours"].corr(
    df["Marks"],
    method="pearson"
)

print("Pearson Correlation:", pearson)
```

# Numerical Example
| Hours  |    Marks |
|---:       |---:|
| 1         | 40 |
| 2         | 50 |
| 3         | 60 |
| 4         | 70 |
| 5         | 80 |

```python
import pandas as pd
import numpy as np

data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80]
}
df = pd.DataFrame(data)
# Pearson correlation
correlation = df["Hours"].corr(df["Marks"])
print("Correlation:", correlation)
```

# 6.Spearman Correlation

# Concept Explanation
Spearman correlation is a rank-based correlation measure.
It is useful when:
- Data is ordinal/ranked
- Relationship is monotonic
- Data does not satisfy the assumptions needed for Pearson correlation
- Outliers make Pearson less suitable

# Formula
Where:
- d_i = difference between ranks
- n = number of observations

```python
spearman = df["Hours"].corr(
    df["Marks"],
    method="spearman"
)
print("Spearman Correlation:", spearman)
```

# 7 Correlation Matrix

```python
# Under stand the Concept
A correlation matrix shows the correlation between every pair of numerical variables.
```

```python
# Example:
correlation_matrix = df.corr(numeric_only=True)
print(correlation_matrix)
```

```python
# Heatmap
import seaborn as sns
import pandas as ps
import matplotlib .pyplot as plt
heatmap = plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```
