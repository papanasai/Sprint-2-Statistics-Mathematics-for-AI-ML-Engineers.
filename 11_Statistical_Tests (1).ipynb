#  – Statistical Tests**

## **1. T-Test**



**Understand the Concept**

* T-test is used to compare average values.
* It helps us check whether the difference between averages is significant.
* It is mainly used when the sample size is small and the population standard deviation is not known.

**Assumptions**

* Data should be numerical.
* The observations should be independent.
* Data should be approximately normally distributed.
* Two groups should be independent.

**When to Use**

* To compare the average of one group with a known value.
* To compare the averages of two groups.

**When Not to Use**

* When the data is categorical.
* When the data has strong outliers.
* When the assumptions of the test are not satisfied.



**Demonstrate the Concept**

**Example:**

* A company wants to check whether the average delivery time is different from 30 minutes.
* A t-test can be used to compare the sample delivery times with 30 minutes.



```python
# Implement the Concept:
import numpy as np
from scipy.stats import ttest_1samp

# Delivery time
delivery_times = np.array([28, 31, 29, 32, 30, 27, 31, 29, 30, 28])

# H0: Average delivery time is 30 minutes
# H1 : Average delivery time is not 30 minutes
result = ttest_1samp(delivery_times, popmean=30)

print("Average Delivery Time:", np.mean(delivery_times))
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)

if result.pvalue < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **2. Z-Test**



**Understand the Concept**

* Z-test is used to compare a sample mean with a known population mean.
* It is mainly used when the population standard deviation is known.
* It is also commonly used for large samples.

**Assumptions**

* Data should be numerical.
* Observations should be independent.
* Population standard deviation should be known.

**When to Use**

* To compare a sample mean with a known population mean.
* To test a mean when the population standard deviation is known.

**When Not to Use**

* When the population standard deviation is unknown and the sample is small.
* For categorical data.
* When the observations are not independent.



**Demonstrate the Concept**

**Example:**

* A packet of biscuits is expected to weigh 100 grams.
* We check the weight of many packets.
* A Z-test can help us check whether the average weight is different from 100 grams.



```python
# Implement the Concept:
from statsmodels.stats.weightstats import ztest

# Biscuits weights 
weights = np.array([498, 501, 502, 499, 500, 497, 503, 501, 499, 500])

# Mean
mean = weights.mean()

# Z-Test
z_stat,p_value = ztest(weights,value=mean)

print("Mean:", mean)
print("Z-score:", z_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **3. Chi-Square Test**



**Understand the Concept**

* Chi-Square test is used with categorical data.
* It helps us check whether two categorical variables are related.
* It compares the observed values with the expected values.

**Assumptions**

* Data should be categorical.
* Observations should be independent.
* The expected count in each category should be large enough.

**When to Use**

* To check the relationship between two categorical variables.

**When Not to Use**

* For numerical data like height or salary.
* When the observations are not independent.
* When expected counts are very small.



**Demonstrate the Concept**

**Example:**

* A shop wants to check whether customers prefer different payment methods.
* It records whether customers use Cash, Card, or UPI.
* A Chi-Square test can be used to compare the observed counts with the expected counts.



```python
# Implement the Concept:
from scipy.stats import chisquare

# No of customers - Using Cash, Card, UPI
observed = np.array([40, 35, 25])

# Expected counts - if all methods were equally preferred
expected = np.array([100 / 3, 100 / 3, 100 / 3])

# H₀: There is no significant difference in the payment preferences.
# H₁: There is a significant difference in the payment preferences.
    
# Chi-Square test
chi_stat, p_value = chisquare(observed, f_exp=expected)

print("Chi-Square Statistic:", chi_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **4. ANOVA**



**Understand the Concept**

* ANOVA means Analysis of Variance.
* It is used to compare the average values of three or more groups.
* It helps us check whether there is a significant difference between the group means.

**Assumptions**

* Data should be numerical.
* Groups should be independent.
* Data should be approximately normally distributed.
* The groups should have similar variance.

**When to Use**

* When we want to compare the means of three or more groups.

**When Not to Use**

* When there are only two groups, a t-test is usually enough.
* When the data is strongly non-normal or has serious outliers.
* When the groups are not independent.



**Demonstrate the Concept**


**Business Example:**

* A company wants to compare the average sales from three different branches.
* ANOVA can check whether the branches have significantly different average sales.



```python
# Implement the Concept:

from scipy.stats import f_oneway

# Sales - 3 branches
branch_a = [20, 22, 21, 19, 23]
branch_b = [25, 27, 26, 24, 28]
branch_c = [30, 31, 29, 32, 30]

# ANOVA test
f_stat, p_value = f_oneway(branch_a, branch_b, branch_c)

print("F-statistic:", f_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **5. Mann-Whitney U Test**



**Understand the Concept**

* Mann-Whitney U test is used to compare two independent groups.
* It checks whether the values in one group are generally higher or lower than the other group.
* It is useful when the data does not follow a normal distribution.

**Assumptions**

* There should be two independent groups.
* Data should be numerical or ordinal.
* The observations should be independent.
* The data does not need to follow a normal distribution.

**When to Use**

* When comparing two independent groups.
* When the data is not normally distributed.

**When Not to Use**

* When the two groups are related or paired.
* When there are more than two groups.
* When the observations are not independent.




**Demonstrate the Concept**


**Example:**

* A company compares customer waiting times between two branches.
* If the waiting-time data is not normally distributed, the Mann-Whitney U test can be used.


```python
# Implement the Concept:
from scipy.stats import mannwhitneyu

# waiting time 
branch_a = [10, 12, 15, 11, 13, 14]
branch_b = [18, 20, 17, 22, 19, 21]

# Mann-Whitney U test
u_stat, p_value = mannwhitneyu(branch_a, branch_b, alternative="two-sided")

print("U-statistic:", u_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Do not Reject the Null Hypothesis")
```

## **6. Shapiro-Wilk Test**



**Understand the Concept**

* Shapiro-Wilk test checks whether the data follows a normal distribution.
* It is mainly used for checking the normality of small datasets.


**Assumptions**

* Data should be numerical.
* Observations should be independent.
* It is mainly useful for checking normality.
* It works best with small to moderate sample sizes.

**When to Use**

* When we want to check whether data is approximately normally distributed.
* It can be used before tests like the t-test or ANOVA.

**When Not to Use**

* It is not used to compare two groups.
* It is not used to check relationships between variables.
* For very large datasets, the test can become overly sensitive to small departures from normality.



**Demonstrate the Concept**

**Example:**

* A company records the delivery time of orders.
* Before using a t-test, the company can check whether the delivery times are approximately normal.



```python
# Implement the Concept:
from scipy.stats import shapiro

# Delivery times 
delivery_times = np.array([28, 31, 29, 30, 32, 27, 30, 29, 31, 28])

# Shapiro-Wilk test
statistic, p_value = shapiro(delivery_times)

print("Shapiro-Wilk Statistic:", statistic)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("Data is not normally distributed")
else:
    print("Do not Reject the Null Hypothesis")
    print("Data is approximately normally distributed")
```

## Interview Questions & Answers

**Q1. What is a T-Test?**

**Answer:**

T-test is used to compare means, especially when the sample is small and population standard deviation is not known.

**Q2. What is a Z-Test?**

**Answer:**

Z-test is used to test a mean when the population standard deviation is known or when a large-sample normal approximation is appropriate.

**Q3. What is a Chi-Square Test?**

**Answer:**

Chi-Square test is used mainly with categorical data to check relationships or compare observed and expected counts.

**Q4. What is ANOVA?**

**Answer:**

ANOVA is used to compare the means of three or more groups.

**Q5. What is Mann-Whitney U Test?**

**Answer:**

Mann-Whitney U test compares two independent groups when the data is not suitable for a t-test.

**Q6. What is Shapiro-Wilk Test?**

**Answer:**

Shapiro-Wilk test checks whether numerical data is approximately normally distributed.

**Q7. How do we interpret p-value in these tests?**

**Answer:**

If p-value is less than 0.05, we usually reject the null hypothesis. Otherwise, we do not reject it.
