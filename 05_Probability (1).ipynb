## **1. Introduction to Probability**

**Understand the Concept**

* Probability tells us how likely something is to happen.
* Its value is between 0 & 1.
* 0 - event cannot happen.
* 1 - event will definitely happen.

**Formula :** P(E) = Number of Favorable Outcomes / Total Number of Outcomes

**Demonstrate the Concept**

**Example:**

* A dice has 6 possible outcomes: 1, 2, 3, 4, 5, 6.
* We want to find the probability of getting 6.
* Favorable outcomes = 1
* Total outcomes = 6
* Probability = 1 / 6 = 0.1667

```python
# Implement the Concept:
# Dice - 6 possible outcomes
total_outcomes = 6

# 1 favorable outcome 
favorable_outcomes = 1

# Probability 
probability = favorable_outcomes / total_outcomes

print("Total Outcomes:", total_outcomes)
print("Favorable Outcomes:", favorable_outcomes)
print("Probability of getting 6:", probability)
```

So, the probability of getting 6 is 16.67%.

## **4. Independent Events**

**Understand the Concept**

* Independent events are events where one event does not affect the other event.
* The result of one event does not change the probability of the second event.

**Formula :** P(A and B) = P(A) × P(B)

**Demonstrate the Concept**

**Example:**

* We toss a coin and roll a dice.
* Getting Head on the coin does not affect the result of the dice.
* Probability of Head = 1 / 2
* Probability of getting 6 = 1 / 6
* Probability of getting Head and 6 = 1 / 2 × 1 / 6 = 1 / 12

```python
# Implement the Concept:
# 2 events
probability_head = 1 / 2
probability_six = 1 / 6

# Probability
probability_both = probability_head * probability_six

print("Probability of Head:", probability_head)
print("Probability of 6:", probability_six)
print("Probability of Head and 6:", probability_both)
```

The probability of getting Head and 6 is 8.33%.

## **5. Dependent Events**

**Understand the Concept**

* Dependent events are events where one event affects the other event.
* The result of the first event changes the probability of the second event.

**Formula :** P(A and B) = P(A) × P(B | A)

**Demonstrate the Concept**

**Example:**

* A bag has 4 red apples and 6 green apples.
* We pick one red apple and do not put it back.
* Now, 3 red apples are left out of 9 apples.
* So, the probability of picking a red apple the second time changes.

```python
# Implement the Concept:
# Apples in bag
red_apples = 4
green_apples = 6
total_apples = 10

# 1st event - pick one red apple
first_event = red_apples / total_apples  # P(A)

# Now 1 red apple is removed in bag
current_red_apples = red_apples - 1 # 3 r apple
current_total_apples = total_apples - 1 # 9 tot apple

# 2nd event - pick another 1 red apple
second_event = current_red_apples / current_total_apples # P(B|A)

# Dependent event 
probability_1_and_2 = first_event * second_event # P(A and B) 

print("1st event probability : ",first_event )
print("2nd event probability of both event : ",second_event )
print("probability of both event : ",probability_1_and_2)
```

The probability of picking two red apples is 13.33%.

## **6. Conditional Probability**

**Understand the Concept**

* Conditional probability tells us the probability of an event when we already know that another event has happened.
* It helps us find the chance of one event under a given condition.

**Formula :** P(A | B) = P(A and B) / P(B)



**Demonstrate the Concept**

**Example:**

* A class has 20 students.
* 12 students play cricket.
* 8 students play both cricket and football.
* We want to find the probability that a student plays football given that the student plays cricket.
* P(Football | Cricket) = 8 / 12 = 0.67


```python
# Implement the Concept:
# Students play cricket
cricket_students = 12

# Play both cricket and football
both_sports = 8

# Conditional probability
probability_football_given_cricket = both_sports / cricket_students
print("P(Football | Cricket):", probability_football_given_cricket)
```

So, the probability of playing football given that the student plays cricket is 66.77%.

## **7. Bayes' Theorem**

**Understand the Concept**

* Bayes' Theorem helps us find the probability of an event using new information.
* It updates an earlier probability when we get new evidence.
* It is useful when we know the result of one event and want to find the chance of another event.

**Formula :** P(A | B) = [P(B | A) × P(A)] / P(B)



**Demonstrate the Concept**

**Example:**

* Suppose 30% of emails are spam.
* 80% of spam emails contain the word "Offer".
* 20% of normal emails also contain the word "Offer".
* We want to find the probability that an email is spam when it contains the word "Offer".

```python
# Implement the Concept:
# Bayes' Theorem:  P(Spam | Offer) = (P(Offer | Spam) * P(Spam)) / P(Offer)

p_spam = 0.30
p_offer_spam = 0.80
p_offer_normal = 0.20
p_normal = 1 - p_spam

# Probability of getting "Offer" in both spam & normal
p_offer = (p_offer_spam * p_spam + p_offer_normal * p_normal)

# Bayes' Theorem
p_spam_given_offer = (p_offer_spam * p_spam) / p_offer

print("P(Offer):", p_offer)
print("P(Spam | Offer):", p_spam_given_offer)
```

* The probability of getting the word "Offer" is 38%.
* If an email contains the word "Offer", the chance that it is spam is about 63.16%.

## **8. Probability Rules**

**Understand the Concept**

* Probability follows some basic rules.
* The probability of an event is always between 0 and 1.
* The probability of the complete sample space is 1.
* The probability of an event not happening is 1 minus its probability.

**Basic Rules**

**Rule 1:** 0 ≤ P(E) ≤ 1

**Rule 2:** P(Sample Space) = 1

**Rule 3:** P(Not A) = 1 - P(A)



**Demonstrate the Concept**

**Example:**

* We roll a dice once.
* Even numbers are 2, 4, and 6.
* Probability of getting an even number = 3/6 = 0.5.
* Probability of not getting an even number = 1 - 0.5 = 0.5.



```python
# Implement the Concept:
total_outcomes = 6
even_outcomes = 3

# Probability of getting an even number
p_even = even_outcomes / total_outcomes

# Probability of not getting an even number
p_not_even = 1 - p_even

print("P(Even):", p_even)
print("P(Not Even):", p_not_even)
print("P(Sample Space):", 1)
```

## **10. Expected Value**

**Understand the Concept**

* Expected value is the average result we can expect from a random event over many times.

**Formula :** Expected Value = Σ (Value × Probability)



**Demonstrate the Concept**

**Example:**

* A game gives ₹100 for a win and ₹0 for a loss.
* Probability of winning = 0.5
* Probability of losing = 0.5

Expected Value = (100 × 0.5) + (0 × 0.5)
Expected Value = ₹50


```python
# Implement the Concept:
# Values
win_amount = 100
loss_amount = 0

# Probability
win_probability = 1/2
loss_probability = 1/2

# Formula : Expected Value = Σ (Value × Probability)
expected_value = (win_amount * win_probability) + (loss_amount * loss_probability)

print("Expected Value:", expected_value)
```

**Explaination :**

* The expected value is ₹50.
* This does not mean we will always win ₹50.
* It means that over many similar games, the average result would be around ₹50 per game.
