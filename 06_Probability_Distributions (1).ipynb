## **1. Bernoulli Distribution**

**Understand the Concept**

* Bernoulli distribution is used when an experiment has only two possible results.
* One result is usually called success and the other is failure.
* Examples are Yes/No, True/False, Pass/Fail.

**Formula :** P(X = x) = pˣ(1 - p)¹⁻ˣ

* x = 0 or 1
* p = probability of success
* 1 - p = probability of failure

**Where it is Used**

* Coin toss
* Product is defective or not
* Customer buys or does not buy
* Email is spam or not spam

**Demonstrate the Concept**

**Example:**

* A customer either buys a product or does not buy it.
* Probability of buying = 0.6
* Probability of not buying = 0.4
* These are the two possible outcomes.

```python
# Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Implement the Concept:
# Probability of buying 
p = 0.6

# Possible outcomes
outcomes = [0, 1]

# Probability 
probabilities = [1 - p, p]

print("Probability of No Buy:", probabilities[0]) # [0] - index
print("Probability of Buy:", probabilities[1])

# Visualization
plt.bar(outcomes, probabilities)
plt.xticks([0, 1], ["No Buy", "Buy"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

## **2. Binomial Distribution**



**Understand the Concept**

* Binomial distribution is used when we repeat the same experiment a fixed number of times.
* Each trial has only two possible results, like Success or Failure.

**Formula :** P(X = k) = C(n, k) × pᵏ × (1 - p)ⁿ⁻ᵏ

* n = number of trials
* k = number of successes
* p = probability of success

**Where it is Used**

* No of customers who buy a product
* No of successful calls
* No of defective products
* No of patients who respond to treatment



**Demonstrate the Concept**

**Example:**

* A coin is tossed 5 times.
* We want to find the probability of getting exactly 3 Heads.
* Number of trials = 5
* Probability of Head = 0.5


```python
# Implement the Concept:
heads = np.random.binomial(n=5, p=0.5)
print("No of Heads:", heads)
```

The result can change each time because the tosses are random.

## **3. Poisson Distribution**



**Understand the Concept**

* Poisson distribution is used to find how many times an event happens in a fixed period or place.
* It is useful for counting events.

**Formula :** P(X = k) = (e⁻λ × λᵏ) / k!

* λ = average number of events
* k = number of events we want to find

**Where it is Used**

* Number of customers arriving at a shop
* Number of calls received by a company
* Number of website visits
* Number of accidents in a day



**Demonstrate the Concept**

**Example:**

* A shop receives an average of 4 customers per hour.
* We want to find the no of customers arriving in 1 hour.
* Average rate = 4 customers per hour.


```python
# Implement the Concept:
# Average 4 customers arrive per hour
customers = np.random.poisson(lam=4)
print("Customers arrived in one hour:", customers)
```

```python
data = np.random.poisson(lam=4, size=1000)

plt.hist(data, bins=np.arange(0, 12),edgecolor="black" )
plt.xlabel("No of Customers")
plt.ylabel("Frequency")
plt.title("Poisson Distribution")
plt.show()
```

## **4. Uniform Distribution**



**Understand the Concept**

* Uniform distribution means all values in a given range have the same chance of occurring.
* The values are spread evenly across the range.
* It can be either discrete or continuous.

**Formula :** P(a ≤ X ≤ b) = 1 / (b - a)

* a = minimum value
* b = maximum value

**Where it is Used**

* Random number generation
* Equal chance selection
* Simulation
* Simple probability experiments



**Demonstrate the Concept**

**Numerical Example:**

* A random number is selected between 1 to 10.
* Every value in this range has the same chance of being selected.
* Minimum value = 1
* Maximum value = 10


```python
# Implement the Concept:
# random value between 1 to 10
value = int(np.random.uniform(1, 10))

print("Random Value:", value)
```

```python
data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=10, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

The values have almost equal frequency

## **5. Normal Distribution**

**Understand the Concept**

* Normal distribution is a bell-shaped distribution.
* Most values are close to the mean.
* Very high and very low values are less common.

**Formula :** f(x) = (1 / (σ√2π)) × e^(-(x-μ)² / 2σ²)

* μ = Mean
* σ = Standard Deviation

**Where it is Used**

* Student marks
* Heights and weights
* Measurement errors
* Statistical analysis

**Demonstrate the Concept**

**Numerical Example:**

* A class has an average mark of 70.
* The standard deviation is 10.
* Most marks will be close to 70.
* Marks far away from 70 will be less common.



```python
# Implement the Concept:

# marks data - mean 70, std 10
marks = np.round(np.random.normal(loc=70, scale=10, size=10),2)

print("Exam Marks:", marks)
print("Mean:", np.mean(marks))
print("Standard Deviation:", np.std(marks))
```

```python
import seaborn as sns
```

```python
marks = np.random.normal(loc=70, scale=10, size=1000)

sns.histplot(data = marks, bins=30, kde = True, color = "orange")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Normal Distribution of Exam Marks")
plt.show()
```

## **6. Exponential Distribution**

**Understand the Concept**

* Exponential distribution is used to measure the time between two events.
* It is mainly used when we want to know how long we need to wait for the next event.

**Formula :** f(x) = λe⁻λˣ

* λ = rate of events
* x = waiting time

**Where it is Used**

* Waiting time for a customer
* Time between phone calls
* Time between website requests
* Time between machine failures



**Demonstrate the Concept**

**Numerical Example:**

* A help desk receives an average of 2 calls per hour.
* We want to simulate the waiting time until the next call.




```python
# Implement the Concept:
# Average 2 calls per hour
calls_per_hour = 2

# Waiting time for the next call
waiting_time = np.random.exponential(scale=1)/2

print("Waiting time for next call is ", waiting_time, "hours")
```

```python
# Visualization
waiting_times = np.random.exponential(scale=0.5, size=1000)

plt.hist(waiting_times, bins=30)
plt.xlabel("Waiting Time (hours)")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")
plt.show()
```

* Most waiting times are short.
* A few waiting times can be much longer.
