Simple Data Set

```python
import pandas as pd

data = {
    "Experience": [1, 2, 3, 4, 5, 6, 7],
    "Age": [25, 35, 45, 65, 75, 85, 95],
    "Salary": [20000, 30000, 40000, 50000, 60000, 70000, 80000],
    "Department": ["HR", "HR", "Sales", "Finance", "Product", "IT", "IT"]
}

df = pd.DataFrame(data)
print(df)
```

# Histogram

# Python implementation

```python
import matplotlib.pyplot as plt
plt.hist(df["Salary"],bins=5, edgecolor="black")
plt.title("Salary distrubtion")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

# 2. Bar Chart

A Bar Chart compares values across different categories.

# When to Use:
Comparing departments.
Comparing sales, profits, or counts.

# Advantages:
Easy comparison.
Suitable for categorical data.
Limitations:

Limitations:
Not suitable for continuous data.

```python
import matplotlib.pyplot as plt
dept = df["Department"].value_counts()

dept.plot(kind="bar")

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

# 3. Line Chart
Shows trends over time or sequential data.

# When to Use:
Monthly sales
Stock prices
Temperature changes

# Advantages:
Clearly shows trends.
Good for time-series data.

# Limitations:
Not suitable for categorical comparisons.

```python
import matplotlib.pyplot as plt
plt.plot(df["Experience"], df["Salary"], marker="o")

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

# 4. Scatter Plot
Shows the relationship between two numerical variables.

```python
# When to Use:
Finding correlation.
Detecting outliers.
```

# Advantages:
Easy to identify relationships.
Shows clusters.

```python
plt.scatter(df["Experience"], df["Salary"])
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```

# 5. Pie Chart
Shows the percentage contribution of categories.

# When to Use:
Showing proportions.
Market share.
Department distribution.

```python
# Advantages:
Easy to understand percentages.
```

# Limitations:
Not suitable when many categories exist.

```python
dept = df["Department"].value_counts()
plt.pie(dept,
        labels=dept.index,
        autopct="%1.1f%%")
plt.title("Department")
plt.show()
```

# 6. Density Plot
Shows the probability distribution of numerical data using a smooth curve.

# When to Use:
Understanding data distribution.
Comparing distributions.

# Advantages:
Smoother than a histogram.
Easy to compare datasets.

# Limitations:
Not suitable for very small datasets.

```python
df["Salary"].plot(kind="density")

plt.title("Salary density plot")
plt.show()
```

# 7. Box Plot
Shows data spread, quartiles, median, and outliers.

# When to Use:
Outlier detection.
Comparing distributions.

# Advantages:
Clearly identifies outliers.
Summarizes data effectively.

# Limitations:
Does not show the exact distribution shape.

```python
plt.boxplot(df["Salary"])
plt.title("Salary Box Plot")
plt.show()
```

# 8. Violin Plot
Combines a box plot and a density plot to show both summary statistics and data distribution.

# When to Use:
Comparing distributions across groups.
Visualizing the shape of the data.

# Advantages:
Shows both density and quartiles.
More informative than a box plot.

# Limitations:
Slightly harder to interpret for beginners.
Requires the seaborn library.

```python
!pip install seaborn 
```

```python
import seaborn as sns
sns.violinplot(y=df['Salary'])
plt.title("Salary Violin Plot")
plt.show()
```

# 9. Pair Plot
Shows pairwise relationships between multiple numerical features.

# When to Use:
Exploratory Data Analysis (EDA).
Finding correlations.

# Advantages:
Displays relationships between all numerical variables.
Useful before building machine learning models.

# Limitations:
Can become cluttered with many features.
Requires the seaborn library.

```python
import seaborn as sns
sns.pairplot(df)
plt.show()
```

# 10. HeatMap
Visualizes the correlation matrix using colors.

# When to Use:
Finding highly correlated features.
Detecting multicollinearity.

# Advantages:
Easy to interpret correlations.
Useful for feature selection.

# Limitations:
Only works with numerical data.

```python
import seaborn as sns
corr = df.corr(numeric_only = True)
sns.heatmap(corr,
            annot=True,
            cmap="viridis")
plt.title('corrleation heatmap')
plt.show()
```
